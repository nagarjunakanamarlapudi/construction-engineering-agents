# 04 — Agentic RAG
**Agentic RAG** lets the question router choose and sequence read-only tools. The visible trace is a structured execution summary; it deliberately does not expose private chain-of-thought.

```python
from civil_copilot.agents.state import ChatRequest
from civil_copilot.data.loaders import load_corpus
from civil_copilot.demo import build_offline_workflow
from civil_copilot.paths import repo_root

workflow = build_offline_workflow(load_corpus(repo_root()))
question = 'Why was ACT-STEEL-009 blocked, what changed, and what evidence closes the issue?'
response = workflow.invoke(ChatRequest(question=question))
```

```python
{'route': response.route, 'trace': [(event.stage, event.title, event.summary) for event in response.trace], 'citations': [c.record_id for c in response.citations]}
```
