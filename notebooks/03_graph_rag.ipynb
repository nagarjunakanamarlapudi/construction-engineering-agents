# 03 — Graph RAG
**Graph RAG** retrieves explicit project relationships as well as text. Here we follow the verified links leaving RFI-087 instead of asking a language model to guess what depends on it.

```python
from civil_copilot.data.loaders import load_corpus
from civil_copilot.graph.service import ProjectGraphService
from civil_copilot.paths import repo_root

corpus = load_corpus(repo_root())
graph = ProjectGraphService(corpus.records, corpus.relationships)
paths = graph.find_paths('RFI-087', max_depth=2, direction='outgoing')
```

```python
[{'path': [node.record_id for node in path.nodes], 'relations': [edge.relationship_type for edge in path.edges], 'provenance': [edge.provenance for edge in path.edges]} for path in paths[:10]]
```
