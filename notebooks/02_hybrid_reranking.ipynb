# 02 — Hybrid retrieval and reranking
**Hybrid retrieval** combines keyword matching (good for IDs such as RFI-087) with semantic similarity (good for meaning). **Reranking** reorders the candidates using exact IDs, revision status, and question overlap.

```python
from civil_copilot.data.loaders import load_corpus
from civil_copilot.demo import build_offline_workflow
from civil_copilot.retrieval.query import QueryContext
from civil_copilot.paths import repo_root

corpus = load_corpus(repo_root())
workflow = build_offline_workflow(corpus)
packet = workflow.tools.retriever.retrieve(QueryContext(question='What changed in RFI-087 and S-204 Rev 5?'))
```

```python
[(item.chunk.record_id, round(item.fused_score, 4), round(item.rerank_score, 4), item.reasons) for item in packet.evidence]
```
